# Week 3 — The Cleaning Sprint

This week we handle missing values and detect outliers. Our cleaning philosophy for this dataset is conservative: we retain all rows, we replace missing values with meaningful defaults, and we detect outliers using the IQR method but keep them in place because they represent real high-value customers.

## Missing Value Strategy

Product_Category_2 and Product_Category_3 are missing when a product does not have a second or third category. Replacing these with 0 is the correct choice — it does not introduce any false information, and it signals to the model that no additional category was purchased.

---

```python
# Make a copy of the original DataFrame so we never modify the raw data
df_clean = df.copy()
```

```python
# Replace missing values in Product_Category_2 with 0
# (0 means the product has no secondary category — not an error)
df_clean['Product_Category_2'] = df_clean['Product_Category_2'].fillna(0)
```

```python
# Replace missing values in Product_Category_3 with 0
# (same logic — 0 = product has no third category)
df_clean['Product_Category_3'] = df_clean['Product_Category_3'].fillna(0)
```

```python
# Verify that all missing values are now filled (should show all zeros)
df_clean.isnull().sum()
```

All missing values have been filled. No rows were dropped. Every transaction record is preserved for analysis.

## Outlier Detection using the IQR Method

We use the IQR (Interquartile Range) method to detect outliers in the Purchase column. The IQR method is preferred over Z-score for this dataset because Purchase is right-skewed — it is not normally distributed — and the Z-score assumes normality.

The IQR method defines outliers as values below Q1 - 1.5 * IQR or above Q3 + 1.5 * IQR. We will detect these, show them in a boxplot, but keep all rows intact because high spenders are real customers, not data errors.


```python
# Detect outliers in the Purchase column using the IQR method:
# Q1 = 25th percentile, Q3 = 75th percentile, IQR = Q3 - Q1
# Outliers are anything below Q1 - 1.5*IQR or above Q3 + 1.5*IQR
Q1 = df_clean['Purchase'].quantile(0.25)
Q3 = df_clean['Purchase'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
```

```python
# Count how many transactions are flagged as statistical outliers
outliers = df_clean[(df_clean['Purchase'] < lower_bound) | (df_clean['Purchase'] > upper_bound)]
outliers.shape[0]
```

The IQR method flagged a number of rows as statistical outliers in the Purchase column. Before deciding what to do, let us visualise the distribution.

```python
# Draw a boxplot and histogram to visualize the Purchase distribution
# Red dashed line = IQR upper bound (outlier cutoff)
# Green dashed line = IQR lower bound
# Dots above the red line = outlier transactions (real high-value customers we keep them)
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Boxplot
axes[0].boxplot(df_clean['Purchase'], patch_artist=True,
                boxprops=dict(facecolor="#4858CF", color="#7872B8"),
                medianprops=dict(color='#2ECC71', linewidth=2.5),
                whiskerprops=dict(color='#2ECC71'),
                capprops=dict(color='#2ECC71'),
                flierprops=dict(marker='o', color='#FF6B6B', alpha=0.3, markersize=3))
axes[0].axhline(upper_bound, color='#E74C3C', linestyle='--', linewidth=1.5, label=f'IQR Upper = Rs {upper_bound:,.0f}')
axes[0].axhline(lower_bound, color='#27AE60', linestyle='--', linewidth=1.5, label=f'IQR Lower = Rs {lower_bound:,.0f}')
axes[0].set_title('Purchase Outliers — IQR Boundaries')
axes[0].set_ylabel('Purchase Amount (Rs)')
axes[0].legend(fontsize=8)

# Histogram with bounds
axes[1].hist(df_clean['Purchase'], bins=60, color="#493FA3", edgecolor='white', alpha=0.85)
axes[1].axvline(upper_bound, color='#E74C3C', linestyle='--', linewidth=1.5, label=f'IQR Upper Bound')
axes[1].axvline(lower_bound, linestyle='--', linewidth=1.5, label=f'IQR Lower Bound')
axes[1].set_title('Purchase Distribution with IQR Boundaries')
axes[1].set_xlabel('Purchase Amount (Rs)')
axes[1].set_ylabel('Frequency')
axes[1].legend(fontsize=8)

plt.suptitle('Week 3 — Outlier Detection using IQR Method', fontsize=13, y=1.02)
plt.tight_layout()
plt.show()
```

The boxplot and histogram confirm that the Purchase column is right-skewed. The red dashed line marks the IQR upper boundary. Points above it are classified as statistical outliers.

However, we are keeping all these records. A customer who spends Rs 20,000 on Black Friday is a real, valuable customer — not a data error. Removing these rows would mean planning the logistics with no preparation for the highest-spending segment.

The logistics implication is the opposite: these outlier customers need the most attention. They are placing the heaviest, most valuable orders and need priority fulfilment.

We detect the outliers, we document them, and we move forward with all 550,068 rows intact.

## Encoding Categorical Columns

```python
# Encode Gender as a number: Female=0, Male=1
# Simple binary mapping since there's no inherent order
df_eng['Gender_enc'] = df_eng['Gender'].map({'F': 0, 'M': 1})
```

Gender is a binary category with no inherent order. We use manual label encoding: F = 0, M = 1. This avoids creating unnecessary extra columns (which One-Hot Encoding would do) while preserving the distinct values.

```python
# Encode Age groups as ordered integers from 0 (youngest) to 6 (oldest)
# Using ordinal encoding because age groups have a natural order
age_map = {'0-17': 0, '18-25': 1, '26-35': 2, '36-45': 3, '46-50': 4, '51-55': 5, '55+': 6}
df_eng['Age_enc'] = df_eng['Age'].map(age_map)
```

Age has a natural order — 18–25 is genuinely less than 26–35. We use manual ordinal encoding to preserve this sequence. If we used LabelEncoder here, it might assign numbers randomly (alphabetically), breaking the logical order.

```python
# One-Hot Encode City_Category into two binary columns: City_A and City_B
# City_C is dropped — when both City_A=0 and City_B=0, that implies City C
# This avoids multicollinearity (the dummy variable trap)
# City_Category is NOMINAL (no order) — use One-Hot Encoding
city_dummies = pd.get_dummies(df_eng['City_Category'], prefix='City').astype(int)
df_eng = pd.concat([df_eng, city_dummies], axis=1)
# Drop City_C to avoid multicollinearity (dummy variable trap)
# City_A=1,City_B=0 => City A; City_A=0,City_B=1 => City B; City_A=0,City_B=0 => City C
df_eng.drop(columns=['City_C'], inplace=True)
print(df_eng[['City_A','City_B']].head())
```

City_Category (A, B, C) has no meaningful numerical order — City A is not inherently 'less than' City B. Ordinal encoding would introduce a false relationship. We apply **One-Hot Encoding**, creating two binary columns: `City_A` and `City_B`. City C is represented implicitly when both are 0, avoiding the dummy variable trap (multicollinearity). This is the correct encoding for nominal categories.


```python
# Encode years stayed in city as ordered integers: '0'->0, '1'->1, ..., '4+'->4
# Ordinal encoding is appropriate since longer stay = higher value
stay_map = {'0': 0, '1': 1, '2': 2, '3': 3, '4+': 4}
df_eng['Stay_enc'] = df_eng['Stay_In_Current_City_Years'].map(stay_map)
```

Stay_In_Current_City_Years has a natural order from 0 to 4+. We encode it with ordinal integers to preserve this sequence.

## Fixing Purchase Skewness

The Purchase column is right-skewed. Skewed distributions cause forecasting models to overestimate the average — leading to over-stocking of expensive products and under-stocking of medium-priced ones. We apply a Yeo-Johnson transformation to bring the distribution closer to a normal bell curve.

---

```python
# Apply Yeo-Johnson transformation to reduce the right skew in Purchase amounts
# This brings the distribution closer to a normal bell curve for better ML performance
pt = PowerTransformer(method='yeo-johnson')
df_eng['Purchase_yj'] = pt.fit_transform(df_eng[['Purchase']])
```

```python
# Plot Purchase distribution before and after the Yeo-Johnson transformation
# Left = raw skewed data; Right = transformed symmetric data
# Skewness value should drop significantly after transformation
# Plot 1: Before Yeo-Johnson Transformation 
fig1, ax1 = plt.subplots(figsize=(8, 4))
sns.histplot(df_eng['Purchase'], bins=60, kde=True,
             color='#E76F51', edgecolor='white', alpha=0.85, ax=ax1)
ax1.set_title(f'Before Transform  (Skew = {df_eng["Purchase"].skew():.2f})')
ax1.set_xlabel('Purchase Amount (Rs)')
ax1.set_ylabel('Frequency')
plt.tight_layout()
plt.show()

# Plot 2: After Yeo-Johnson Transformation 
fig2, ax2 = plt.subplots(figsize=(8, 4))
sns.histplot(df_eng['Purchase_yj'], bins=60, kde=True,
             color="#1B4F8A", edgecolor='white', alpha=0.85, ax=ax2)
ax2.set_title(f'After Yeo-Johnson  (Skew = {df_eng["Purchase_yj"].skew():.2f})')
ax2.set_xlabel('Transformed Purchase')
ax2.set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```

Before the transformation, the distribution has a long right tail — classic right skew. After Yeo-Johnson, the distribution is nearly symmetric and bell-shaped. The skewness has dropped from around 0.59 toward 0.

This matters for logistics forecasting: a symmetric distribution gives models a better baseline prediction of "typical" spend, which leads to more accurate demand estimates and fewer stock-out or over-order situations on Black Friday.

## Feature Engineering: 7 New Logistics-Relevant Features

This is the core of Week 6. We create 7 new features that each capture a specific pattern relevant to the Black Friday logistics challenge. For each feature, we explain what it is, why we created it, and how it connects to the logistics problem.

---

### Feature 1: category_depth

What it is: The number of product categories a single transaction covers (1, 2, or 3).

Why we created it: A product tagged under three categories is a complex, multi-department item. It may need to be picked from multiple warehouse zones, placed in a larger box, and handled by more than one packer. A product with only one category is simple — pick, pack, ship.

Logistics relevance: category_depth directly tells the warehouse management system how much effort each order requires. High-depth orders should be flagged for multi-zone picking and a bundling station.

```python
# Feature 1 — category_depth:
# Count how many product categories each transaction covers (1, 2, or 3)
# Depth-1 = simple single-zone pick; Depth-3 = complex multi-zone pick
df_eng['category_depth'] = (
    (df_eng['Product_Category_1'] > 0).astype(int) +
    (df_eng['Product_Category_2'] > 0).astype(int) +
    (df_eng['Product_Category_3'] > 0).astype(int)
)
```

---

### Feature 2: is_high_value

What it is: A binary flag (1 = yes, 0 = no) that identifies transactions in the top 20% of spending.

Why we created it: From our EDA we know that a minority of Black Friday customers spend significantly more than the average. These high-value orders are heavier, require more packaging, and may involve premium products. They should not be mixed with standard orders in the fulfilment queue.

Logistics relevance: is_high_value can be fed directly into a warehouse routing system to send the order to an express fulfilment lane with priority dispatch, signature-required delivery, and enhanced packaging standards.

```python
# Feature 2 — is_high_value:
# Flag transactions in the top 20% of spending as 1, everything else as 0
# These orders need priority handling and premium packaging
df_eng['is_high_value'] = (df_eng['Purchase'] >= df_eng['Purchase'].quantile(0.80)).astype(int)
```

---

### Feature 3: is_new_resident

What it is: A binary flag (1 = yes) for customers who have lived in their city for 0 or 1 years.

Why we created it: New residents are a distinct buying segment. They are setting up a new home — which means they are far more likely to buy large items like appliances, furniture, and electronics. These products are bulky and require different packaging and delivery vehicles compared to smaller lifestyle purchases.

Logistics relevance: Knowing that a high proportion of new-resident orders are incoming allows the distribution centre to pre-position trucks with sufficient carrying capacity and to schedule delivery time windows for the morning (when residents are likely to be home).

```python
# Feature 3 — is_new_resident:
# Flag customers who have lived in their city 0 or 1 years (1=yes, 0=no)
# New residents tend to buy larger items: appliances, furniture, home electronics
df_eng['is_new_resident'] = df_eng['Stay_enc'].isin([0, 1]).astype(int)
```

---

### Feature 4: is_young_male

What it is: A binary flag (1 = yes) identifying the core customer segment — male customers aged 18–35.

Why we created it: Our EDA showed that male customers aged 18–35 account for the largest share of transactions. This segment predominantly buys electronics, gadgets, and tech accessories — all of which need specific packaging, voltage compatibility checks, and in many cases warranty-sticker placement.

Logistics relevance: Flagging this segment allows the warehouse to predict surges in electronics zone demand before the day begins. Zone managers can pre-stage additional picking carts, open extra packing stations, and alert the electronics restocking crew first.

```python
# Feature 4 — is_young_male:
# Flag male customers aged 18-35 (1=yes, 0=no)
# This is the dominant buyer segment — primarily purchasing electronics
df_eng['is_young_male'] = ((df_eng['Age_enc'].isin([1, 2])) & (df_eng['Gender_enc'] == 1)).astype(int)
```

---

### Feature 5: spend_per_category

What it is: The purchase amount divided by the category_depth — spend intensity per category level.

Why we created it: Two orders can have the same purchase amount but very different logistics complexity. An order of Rs 10,000 for a single product (category_depth = 1) is simple to fulfil. An order of Rs 10,000 for a 3-category product bundle (category_depth = 3) needs three separate picking operations. This feature normalises spend by complexity.

Logistics relevance: spend_per_category helps prioritise which high-value multi-category orders need the most careful handling. A very high spend_per_category on a depth-1 product suggests a premium single item, while a low value on a depth-3 order suggests three small, low-cost items that are still logistically demanding.

```python
# Feature 5 — spend_per_category:
# Divide purchase amount by category_depth to normalize spend by order complexity
# High value = expensive single item; low value = cheap multi-category bundle
df_eng['spend_per_category'] = df_eng['Purchase'] / df_eng['category_depth']
```

---

### Feature 6: product_bundle_flag

What it is: A binary flag (1 = yes) for products that span two or more categories.

Why we created it: Products that appear in both Category 1 and Category 2 (or all three) are bundle products. They involve more than one picking location in the warehouse, require careful consolidation, and need a larger box or a different packaging format.

Logistics relevance: product_bundle_flag can trigger a bundle-assembly workflow in the warehouse management system, routing the order to a consolidation zone rather than a direct-to-ship zone. Without this flag, bundle orders could be accidentally separated and shipped as two partial deliveries.

```python
# Feature 6 — product_bundle_flag:
# Flag orders that span 2+ product categories (1=bundle, 0=single category)
# Bundle orders need consolidation across multiple warehouse zones before shipping
df_eng['product_bundle_flag'] = (df_eng['category_depth'] >= 2).astype(int)
```

---

### Feature 7: repeat_buyer_score

What it is: A proxy score for "established buyer" behaviour, calculated as Age_enc multiplied by Stay_enc.

Why we created it: A customer who is older AND has lived in their city for several years is likely to have stable, predictable buying habits. They know what they want, they know which categories they buy from, and their Black Friday purchases are likely to be repeats of past behaviour. A young new-resident, by contrast, is exploratory and unpredictable.

Logistics relevance: repeat_buyer_score helps identify orders that are likely to follow predictable category patterns — allowing the logistics team to pre-stock those zones based on historical purchase data for that segment. Lower scores indicate more exploratory buyers, suggesting a wider variety of products may be ordered.

The score is always 0 regardless of age. This means older customers who are new to a city will have the same score as a young new-resident. This edge case is accepted here as a deliberate simplification — the feature still correctly identifies the majority of established-buyer behaviour.

---

```python
# Feature 7 — repeat_buyer_score:
# Multiply age rank by years-in-city to estimate 'established buyer' behaviour
# High score = older long-term resident with predictable buying habits
# Known limitation: score is always 0 when Stay_enc=0 regardless of age
df_eng['repeat_buyer_score'] = df_eng['Age_enc'] * df_eng['Stay_enc']
```

## Correlation Heatmap (Filter Method)

The first step in feature selection is to look at correlations. Highly correlated features (correlation above 0.80) are redundant — they carry the same information twice, which can confuse machine learning models.

---

```python
# Plot a heatmap of correlations between all 17 scaled features
# Features with correlation above 0.80 are candidates for removal (they carry duplicate info)
fig, ax = plt.subplots(figsize=(14, 11))
corr_matrix = X.corr()
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, linewidths=0.4, annot_kws={'size': 7}, ax=ax)
ax.set_title('Feature Correlation Heatmap', fontsize=13)
plt.tight_layout()
plt.show()
```

The heatmap reveals several highly correlated feature pairs. The most notable ones are:

- `is_high_value_sc` is strongly correlated with `Purchase_yj_sc` since high-value customers are defined by their purchase amount.
- `spend_per_category_sc` is closely linked to `Purchase_yj_sc` for similar reasons — both are derived from the Purchase column.
- `product_bundle_flag_sc` and `category_depth_sc` are naturally correlated since bundles always have deeper category depth.

We will use these observations to guide which features to drop.

**Drop decision:** `product_bundle_flag_sc` is dropped because it has a measured pairwise correlation of 0.853 with `category_depth_sc` (above the 0.80 threshold). `is_high_value_sc` and `spend_per_category_sc` are dropped on logical/definitional grounds — both are directly derived from the `Purchase` column, making them conceptually collinear with `Purchase_yj_sc` even if their pairwise r did not breach 0.80 in isolation. These three drops are therefore a combination of the statistical filter and explicit feature-derivation reasoning.

---

```python
# Loop through all feature pairs and collect those with correlation above 0.80
# These highly correlated pairs represent redundant features
# Identify pairs with correlation above 0.80
corr_pairs = []
for i in range(len(corr_matrix.columns)):
    for j in range(i+1, len(corr_matrix.columns)):
        val = abs(corr_matrix.iloc[i, j])
        if val > 0.80:
            corr_pairs.append({
                'Feature A': corr_matrix.columns[i],
                'Feature B': corr_matrix.columns[j],
                'Correlation': round(val, 3)
            })

pd.DataFrame(corr_pairs).sort_values('Correlation', ascending=False)
```

These are the redundant feature pairs. For each pair, we keep the more interpretable or more informative feature and drop the duplicate.

---

```python
# Drop the 3 features identified as redundant from the correlation analysis:
# is_high_value and spend_per_category are derived from Purchase (logically collinear)
# product_bundle_flag has r > 0.80 with category_depth (statistically redundant)
# Drop the redundant features identified from the correlation analysis.
cols_to_drop = ['is_high_value_sc', 'spend_per_category_sc', 'product_bundle_flag_sc']
X_reduced = X.drop(columns=cols_to_drop, errors='ignore')
X_reduced.shape
```

We have reduced from 17 features to **14**. Now we run VIF analysis to check for hidden multicollinearity that the pairwise correlation matrix may have missed.

##  Variance Inflation Factor (VIF)

VIF measures how much each feature is explained by a linear combination of the other features. A VIF above 10 indicates severe multicollinearity — the feature is essentially a near-duplicate of the others combined.



```python
# Helper function to compute VIF (Variance Inflation Factor) for each feature
# VIF > 10 means the feature is nearly explained by a combination of other features
def compute_vif(X_data):
    vif_df = pd.DataFrame()
    vif_df['Feature'] = X_data.columns
    vif_df['VIF'] = [variance_inflation_factor(X_data.values, i) for i in range(X_data.shape[1])]
    return vif_df.sort_values('VIF', ascending=False).reset_index(drop=True)

vif_initial = compute_vif(X_reduced)
vif_initial
```

We now iteratively remove the feature with the highest VIF score (above 10) one at a time until all remaining features are below the threshold.

---

```python
# Iteratively remove the feature with the highest VIF score (above 10)
# Repeat until all remaining features have VIF <= 10 (no severe multicollinearity)
X_vif = X_reduced.copy()
dropped_by_vif = []

while True:
    vif_now = compute_vif(X_vif)
    worst   = vif_now.iloc[0]
    if worst['VIF'] <= 10:
        break
    X_vif = X_vif.drop(columns=[worst['Feature']])
    dropped_by_vif.append(worst['Feature'])

dropped_by_vif
```

```python
# Show the final VIF scores of all remaining features after iterative removal
vif_final = compute_vif(X_vif)
vif_final
```

After iterative VIF elimination, all remaining features have a VIF score below 10. This confirms that the features are sufficiently independent of each other — no feature can be reconstructed by combining the others.

## Variance Thresholding

A feature with near-zero variance provides no predictive signal — every row has essentially the same value, so the feature cannot distinguish one transaction from another. We remove any feature whose variance falls below a threshold of **0.05**.


```python
# Remove any feature whose variance is below 0.05 (near-constant columns)
# A nearly constant feature provides no useful information for predictions
X_final = X_vif.copy()

vt = VarianceThreshold(threshold=0.05)
vt.fit(X_final)

feature_mask         = vt.get_support()
low_variance_dropped = X_final.columns[~feature_mask].tolist()
X_final              = X_final.loc[:, feature_mask]

low_variance_dropped
```

No features were eliminated by variance thresholding. All remaining features have active, meaningful variance. This confirms our earlier selection was sound.

##  PCA: Feature Extraction

PCA transforms our remaining features into a new set of axes (principal components) that capture the maximum variance in the data. Each principal component is a linear combination of the original features, and the components are ordered from most to least variance explained.

---

```python
# Run full PCA on the feature matrix (data is already scaled — do NOT scale again)
# Find how many principal components are needed to explain at least 95% of variance
# X_final is already StandardScaled from Week 6 — do NOT scale again.
# Applying StandardScaler twice is redundant and distorts variance estimates.
X_scaled_pca = X_final.values  # use directly

pca_full = PCA()
pca_full.fit(X_scaled_pca)

exp_var = pca_full.explained_variance_ratio_
cum_var = np.cumsum(exp_var)

n_components_95 = int(np.argmax(cum_var >= 0.95)) + 1
print(f'Components needed to reach 95% variance: {n_components_95}')
n_components_95
```

```python
# Print the exact cumulative variance explained at the 95% cutoff point
print(f'Cumulative variance at {n_components_95} components: {cum_var[n_components_95 - 1]:.4f} ({cum_var[n_components_95 - 1]*100:.2f}%)')
```

PCA identified the minimum number of principal components needed to explain **95% of the total variance** in the data. This is our target dimensionality.

---

## Scree Plot

```python
# Scree plot: show how much variance each individual principal component explains
# Dark bars = retained components; light bars = discarded (contribute < 5% total)
exp_var_pct = exp_var * 100
x_ticks = np.arange(1, len(exp_var) + 1)
colors_bar = ['#1B4F8A' if i <= n_components_95 else '#8ECAE6' for i in x_ticks]

fig, ax = plt.subplots(figsize=(10, 5))
bars = ax.bar(x_ticks, exp_var_pct, color=colors_bar, edgecolor='white', width=0.7)
ax.set_xlabel('Principal Component')
ax.set_ylabel('Variance Explained (%)')
ax.set_title('Scree Plot — Variance Explained by Each Component  (dark = retained)')
ax.set_xticks(x_ticks)
for bar, val in zip(bars, exp_var_pct):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.2,
            f'{val:.1f}%', ha='center', va='bottom', fontsize=8)
plt.tight_layout()
plt.show()
```

The Scree Plot shows how much variance each principal component explains individually. The **dark blue bars** are the components we retain (those needed to reach 95% cumulative variance). The light blue bars are the discarded components — together they contribute less than 5% of the total information.

PC1 carries the largest single share of variance, and each subsequent component carries progressively less. This is the expected "elbow" pattern in a well-formed PCA.

---

##  Cumulative Variance Curve

```python
# Cumulative variance curve showing how quickly components accumulate explanatory power
# Red lines mark the 95% threshold and the exact number of components needed
cum_var_pct = cum_var * 100

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x_ticks, cum_var_pct, color='#1B4F8A', marker='o', linewidth=2.5, markersize=7, label='Cumulative Variance')
ax.fill_between(x_ticks, cum_var_pct, alpha=0.15, color='#1B4F8A')
ax.axhline(y=95, color='#E76F51', linestyle='--', linewidth=1.8, label='95% Threshold')
ax.axvline(x=n_components_95, color='#E76F51', linestyle='--', linewidth=1.8)
ax.annotate(f'{n_components_95} components\n= {cum_var_pct[n_components_95-1]:.1f}%',
            xy=(n_components_95, cum_var_pct[n_components_95-1]),
            xytext=(n_components_95 + 0.5, cum_var_pct[n_components_95-1] - 6),
            fontsize=9, color='#1B4F8A',
            arrowprops=dict(arrowstyle='->', color='#1B4F8A'))
ax.set_xlabel('Number of Components')
ax.set_ylabel('Cumulative Variance Explained (%)')
ax.set_title('Cumulative Variance Explained by PCA Components')
ax.set_xticks(x_ticks)
ax.legend()
plt.tight_layout()
plt.show()
```

The cumulative variance curve shows how quickly the principal components accumulate explanatory power. The curve rises steeply for the first few components, then flattens — confirming that diminishing returns set in quickly and we do not need many components to capture most of the information.

The red dashed lines mark the 95% threshold and the exact number of components needed to cross it. Beyond that point, each additional component adds very little new information.

---

## Applying the Optimised PCA Model

```python
# Apply PCA with exactly n_components_95 components and transform the data
# Each row is now represented as a point in the reduced PCA space
pca_final = PCA(n_components=n_components_95)
X_pca = pca_final.fit_transform(X_scaled_pca)

pca_cols = [f'PC{i}' for i in range(1, n_components_95 + 1)]
df_pca   = pd.DataFrame(X_pca, columns=pca_cols)
df_pca.head()
```

Our original feature matrix has been compressed into a smaller set of uncorrelated principal components. Each PC is a linear combination of the original features, and together they retain over 95% of the variance in the data.
