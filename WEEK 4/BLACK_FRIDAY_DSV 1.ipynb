```python
#Product_Category_2 and Product_Category_3 represent optional product categories
#Missing values indicate that the product does not belong to that additional category
df['Product_Category_2'] = df['Product_Category_2'].fillna(0) #Hence missing values are replaced with 0 instead of removing the rows
df['Product_Category_3'] = df['Product_Category_3'].fillna(0)
```

```python
print("After filling missing values")
df.isnull().sum() 
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df['Purchase'])
plt.title("Outlier Detection — Purchase")
plt.show()
```

```python
df['Purchase'].agg(['min','max'])
```

```python
Q1 = df['Purchase'].quantile(0.25)
Q3 = df['Purchase'].quantile(0.75)
IQR = Q3 - Q1
```

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers_iqr = df[(df['Purchase'] < lower) | (df['Purchase'] > upper)]
```

```python
print("IQR Outlier Count:", outliers_iqr.shape[0])
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
outliers_iqr.head()
```

```python
outliers_iqr['Purchase'].agg(['min','max'])
```

```python
# Some purchase values are very high, but this is normal during Black Friday.
# So we kept the outliers instead of removing them.
plt.figure(figsize=(8,4))
sns.boxplot(x=df['Purchase'])
plt.title("Outlier are retained for Purchase")
plt.show()
```

```python
plt.figure()
top_products = df['Product_ID'].value_counts().head(10).reset_index()
top_products.columns = ['Product_ID', 'Count']

sns.barplot(x='Product_ID', y='Count', data=top_products)
plt.title("Top 10 Most Sold Products")
plt.xticks(rotation=90)
plt.show()
```

```python
#Multivariate analysis
plt.figure(figsize=(10,5))
sns.barplot(x='City_Category', y='Purchase', hue='Product_Category_1', data=df, errorbar=None)
plt.title("Purchase by City and Product Category")
plt.xlabel("City Category")
plt.ylabel("Average Purchase")
plt.legend(title="Product Category", bbox_to_anchor=(1.05,1))
plt.show()
```
